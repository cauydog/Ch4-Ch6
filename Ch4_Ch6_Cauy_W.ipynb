## Chapter 4 — Choose with `if`

### Exercise 4.1 — Compare a guess with a secret number

Choose two numbers between 1 and 10. Use `if`, `elif`, and `else` to tell whether the guess is too low, too high, or correct.

```python
secret = 7
guess = 5

if guess < secret:
    print("too low")
elif guess > secret:
    print("too high")
else:
    print("just right")
```

## Chapter 6 — Loop with `while` and `for`

### Exercise 6.1 — Print every number in a list

Use a `for` loop to print each element in `[3, 2, 1, 0]`.

```python
for number in [3, 2, 1, 0]:
    print(number)
```

**Answer:** The loop runs once for each item in the list and prints `3`, `2`, `1`, and `0` on separate lines.

### Exercise 6.2 — Guess a number with a `while` loop


```python
guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
    number += 1
```

**Answer:** `too low` prints six times while `number` increases from 1 through 6. When it reaches 7, the program prints `found it!` and `break` ends the loop. The `oops` branch handles a value greater than 7.

### Exercise 6.3 — Guess a number with a `for` loop

Set `guess_me` to 5. Loop through `range(10)` and compare `number` with `guess_me`. Print the appropriate message and stop with `break` once the guess is found or passed.

```python
guess_me = 5

for number in range(10):
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
```

**Answer:** The `for` loop checks numbers beginning at 0. It prints `too low` five times for 0 through 4, then prints `found it!` at 5 and exits. `range(10)` supplies the integers from 0 to 9.
